```python
# How to access capture video and camera
import cv2
```

```python
import os
print(os.getgroups())
```

```python
cap = cv2.VideoCapture(1)
i=0
while True:
    flag, frame = cap.read()
    if not flag:
        print("camera not accessible")
        break
    cv2.imshow("Camera", frame)
    i+=1
    cv2.imwrite(f"capture_image/img_{i}.jpg", frame)
    key=cv2.waitKey(25) & 0xFF

    if key==27: #ESC
        break
cap.release()
cv2.destroyAllWindows()
```
