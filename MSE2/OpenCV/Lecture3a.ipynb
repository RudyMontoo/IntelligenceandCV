```python
# FACE DETECTION
-> Haar features algorithms = (color-> black and white->)
```

Haar Feature  https://miro.medium.com/v2/resize:fit:1288/1*SL7QfuqVAY70m-zO41RHuw.png


```python
import cv2
```

```python
haar_features=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
img=cv2.imread('msd.jpeg')
```

```python
haar_features.detectMultiScale(img,scaleFactor=1.3)  # give scale factor is use when it not detcting correctly
```

```python
faces=haar_features.detectMultiScale(img)
for (x,y,w,h) in faces:
    #img- on which we want to draw rectangle
    # (x,y)- tp left coeenr
    # (x+w,y+h) bottom-right corner
    # (0,255,0) green color border
    # 2- border thickness
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)

cv2.imshow('output',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
