```python
import cv2
import yt_dlp

URL = "https://www.youtube.com/watch?v=9FDdQs3AH58"

ydl_options = {
    'format': 'best[ext=mp4][protocol=https]'
}

with yt_dlp.YoutubeDL(ydl_options) as ydl:
    info = ydl.extract_info(URL, download=False)

    video_url = info['url']

cap = cv2.VideoCapture(video_url)

i = 0

if not cap.isOpened():
    print("Failed to open video stream")
    exit()

while True:
    ret, frame = cap.read()

    if not ret:
        print("Stream ended or not accessible")
        break

    cv2.imshow("YouTube Stream", frame)

    # save frames (optional)
    i += 1
    cv2.imwrite(f"capture_image/img_{i}.jpg", frame)

    if cv2.waitKey(25) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
