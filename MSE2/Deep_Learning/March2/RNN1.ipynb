```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, SimpleRNN
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
text=[
    "i am learning deep learning",
    "i like deep learning",
    "deep learning is fun to learn",
    "i also like machine learning"
]
```

```python
# tokenization
tokenizer=Tokenizer()
tokenizer.fit_on_texts(text)
```

```python
sequences=tokenizer.texts_to_sequences(text)
print(sequences)
```

```python
# create input output pair
X=[]
y=[]
for seq in sequences:
    for i in range(1, len(seq)):
        X.append(seq[:i])
        y.append(seq[i])
```

```python
# PADDING
X= pad_sequences(X)
y=np.array(y)
```

```python
print(X)
```

```python
print(y)
```

```python
vocab_size=len(tokenizer.word_index)+1
```

```python
print(vocab_size)
```

```python
# BUILD RNN MODEL

model=Sequential([
    Embedding(vocab_size,10,input_length=X.shape),
    SimpleRNN(32),
    Dense(vocab_size, activation='softmax')
])
```

```python
model.compile(loss="sparse_categorical_crossentropy",optimizer="adam")
```

```python
model.fit(X,y,epochs=100)
```

```python
input_text="i am"
seq=tokenizer.texts_to_sequences([input_text])[0]
seq=X = pad_sequences([seq], maxlen=10, padding='post')
pred=model.predict(seq)
```

```python
print(pred)
```

```python
pred_index=np.argmax(pred)
```

```python
for word, index in tokenizer.word_index.items():
    if index==pred_index:
        print("Next word:",word)
```
