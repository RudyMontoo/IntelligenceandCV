```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
reviews = [ "i like this movie",
            "this movie is great",
            "i do not like this movie",
            "this movie is terrible" 
            ]

labels = [1, 1, 0, 0]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(reviews)
sequences = tokenizer.texts_to_sequences(reviews)
```

```python
print(sequences)
```

```python
X= pad_sequences(sequences)
y=np.array(labels)
```

```python
vocab_size = len(tokenizer.word_index) + 1
max_length = X.shape[1]
```

```python
model = Sequential([Embedding(vocab_size, 100, input_length=max_length),
                    LSTM(32),Dense(1, activation='sigmoid')])
model.compile(optimizer='adam', loss='binary_crossentropy')
model.fit(X, y, epochs=50)
```

```python
test = ["i like this movie", "this movie is terrible"]
test_seq = tokenizer.texts_to_sequences(test)
test_pad = pad_sequences(test_seq, maxlen=max_length)
predictions = model.predict(test_pad)
print(predictions)
```
